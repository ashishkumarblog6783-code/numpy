```python
s=pd.Series(data=["p","w","p","l","k","c"],dtype="category")

print("series",s)

s=s.cat.add_categories("t")
print(s)
```

```python
s=pd.Series(data=["p","o","p","d","m","b"],dtype="category")

print("series=\n",s)

s=s.cat.remove_categories("p")

print("updated\n",s)
```
