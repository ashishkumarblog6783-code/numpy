```python
a = [12, 34, 55, 33, 44]

# Remove duplicates and sort the list in ascending order
# Converting to a set first removes duplicates
unique_sorted_a = sorted(list(set(a)))

# Check if the list has at least two unique elements
if len(unique_sorted_a) < 2:
    print("List does not contain at least two unique elements to find the second largest.")
else:
    # The second largest element will be the second to last after sorting
    second_largest = unique_sorted_a[-2]
    print(f"The second largest element is: {second_largest}")
```
