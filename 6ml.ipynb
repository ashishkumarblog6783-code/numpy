fill between

```python
import matplotlib.pyplot as plt

x=[2,4,5,6]
y=[5,6,7,8]

plt.plot(x,y)

plt.plot(x,y,color="r")
plt.xlabel("x-axis")
plt.fill_between(x,y,color="b")
plt.ylabel("y-axis")
plt.title("fill between")
plt.show()
```
