```python
import matplotlib.pyplot as plt
x=[23,12,55,65,32]
plt.boxplot(x)
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[23,12,55,65,32]
plt.boxplot(x,notch=True)
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[23,12,55,65,32]
plt.boxplot(x,vert=False)
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[23,12,55,65,32]
plt.boxplot(x,widths=0.5)
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[23,12,55,65,32]
plt.boxplot(x,showmeans=True,whis=2)
plt.xlabel('Data')
plt.ylabel('ladla')
plt.title('kya baat ay')
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[23,12,55,65,32]
plt.boxplot(x,sym="r")
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[23,12,55,65,32]
plt.boxplot(x,boxprops=dict(color="r"),capprops=dict(color="g"),flierprops=dict(markeredgecolor="y"))
plt.show()
```
