```python
from IPython.core.display import display_pdf

import pandas as pd
data={
    'id':["s01","s02","s03","s04","s05"],
    'student':["sihu","lojei","gghd","gini","pmi"],
    'rank':[2,3,4,5,6],
    'marks':[67,78,55,34,90]

}
df=pd.DataFrame(data,index=["row1","row2","row3","row4","row5"])

print(df)

print(df.sort_values(by=['rank','marks']))
```

```python

import pandas as pd
data={
    'id':["s01","s02","s03","s04","s05"],
    'student':["sihu","lojei","gghd","gini","pmi"],
    'rank':[2,3,4,5,6],
    'marks':[67,78,55,34,90]

}
dataFrame=pd.DataFrame(data)
print(dataFrame)
print(dataFrame.sort_values(by=["rank"],ascending=False))
```

```python

import pandas as pd
data={
    'id':["s01","s02","s03","s04","s05","s09"],
    'student':["sihu","lojei","gghd","gini","pmi","sihu"],
    'rank':[2,3,4,5,6,9],
    'marks':[67,78,55,34,90,9]

}
dataFrame=pd.DataFrame(data)
print(dataFrame)

res=dataFrame.duplicated()
print(res)
```
