stack plot


```python
import matplotlib.pyplot as plt
x=[1,3,4,5,6]
area=[1,2,3,4,5]
area2=[2,4,3,6,7]
l=["area1","area2"]
plt.xlabel("olllre")
plt.ylabel("pingpong")
plt.legend(l)
plt.stackplot(x,area,area2,colors=["g","r"])
plt.show()
```

step plot

```python
import matplotlib.pyplot as plt
x=[1,3,4,5,6]
y=[2,3,4,5,6]
plt.step(x,y)
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1,3,4,5,6]
y=[2,3,4,5,6]
plt.step(x,y,marker="o",ms=10,mfc="g",label="python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.title("python",color="r")
plt.grid()
plt.show()
```
