When you re-upload a file in Colab, it often appends a number to the filename (e.g., `students.csv (1).xlsx`). To ensure you're always reading the most recent version, you can retrieve the actual filename from the `uploaded` dictionary returned by `files.upload()`.

```python
from google.colab import files
uploaded = files.upload()

import pandas as pd
df = pd.read_excel("students.csv.xlsx",index_col="student")
print(df)


print(df.loc("loki"))
```
