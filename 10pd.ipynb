```python

import pandas as pd
data={
    'id':["s01","s02","s03","s04","s05"],
    'student':["sihu","lojei","gghd","gini","pmi"],
    'rank':[2,3,4,5,6],
    'marks':[67,78,55,34,90]

}
dataFrame=pd.DataFrame(data)
print(dataFrame)

for row in dataFrame.iterrows():
  print(row)
```

```python

import pandas as pd
data={
    'id':["s01","s02","s03","s04","s05"],
    'student':["sihu","lojei","gghd","gini","pmi"],
    'rank':[2,3,4,5,6],
    'marks':[67,78,55,34,90]

}
dataFrame=pd.DataFrame(data)
print(dataFrame)

for a,b in dataFrame.items():
  print(a)
  print(b)
```
