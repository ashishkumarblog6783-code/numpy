```python
import pandas as pd

data={
    'student':["amit","johny","tony","pony","dolly"],
    'rank':[1,2,3,4,5],
    'marks':[89,76,90,23,54]

}

df = pd.DataFrame(data)
print("student record",df)
print("\ndisplaying the column:")
for col in df:
  print(col)
```
